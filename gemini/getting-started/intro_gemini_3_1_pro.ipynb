## 🚀 Quickstart

By default, Gemini 3.1 Pro uses dynamic thinking to reason through prompts. For faster, lower-latency responses when complex reasoning isn't required, you can constrain the model's thinking level by setting parameter `thinking_level` to `low`.  
默认情况下，Gemini 3.1 Pro 会使用动态思考（dynamic thinking）来对提示进行推理。当不需要复杂推理、希望获得更快、延迟更低的响应时，可以通过将参数 `thinking_level` 设置为 `low` 来限制模型的思考级别。

## 🚨 New API Features

### 1️⃣ Thinking Level  
### 1️⃣ 思考级别（Thinking Level）

The `thinking_level` parameter allows you to specify a thinking budget for the model's response generation. By selecting a state, you can explicitly balance the trade-offs between response quality/reasoning complexity and latency/cost.  
`thinking_level` 参数允许您为模型的响应生成指定思考预算。通过选择不同级别，您可以在响应质量/推理复杂度与延迟/成本之间进行权衡。

- **Low**: Minimizes latency and cost. Best for simple instruction following or chat. Maps to a token budget between `1` and `1,000`.  
  **低（Low）**：最小化延迟和成本，适用于简单的指令执行或对话场景。对应的 token 预算范围为 `1` 到 `1,000`。

- **Medium**: A balance between reasoning depth and latency. Maps to a token budget between `1,001` and `16,384`.  
  **中（Medium）**：在推理深度与延迟之间取得平衡。对应的 token 预算范围为 `1,001` 到 `16,384`。

- **High** (default): Maximizes reasoning depth. The model may take significantly longer to reach a first token, but the output will be more thoroughly vetted. Maps to a token budget between `16,384` and `32,768`.  
  **高（High，默认）**：最大化推理深度。模型生成首个 token 的时间可能明显更长，但输出内容会更加严谨。对应的 token 预算范围为 `16,384` 到 `32,768`。

#### ⚠️ Notes  
#### ⚠️ 注意事项

- If `thinking_level` is not specified, the model defaults to `high`, which uses a dynamic token budget based on prompt complexity.  
  如果未指定 `thinking_level`，模型默认使用 `high`，并根据提示复杂度动态分配 token 预算。

- Setting a token budget of `0` will result in an error.  
  将 token 预算设置为 `0` 会导致错误。

- As a mitigation for looping, the model will output a thought block, even if it's empty, for every step in a multi-step interaction.  
  为了避免循环问题，在多步骤交互中，模型会在每一步输出一个思考块（即使为空）。

- You cannot use both `thinking_level` and the legacy `thinking_budget` parameter in the same request. Doing so will return a 400 error.  
  不能在同一请求中同时使用 `thinking_level` 和旧参数 `thinking_budget`，否则会返回 400 错误。

- The OpenAI Chat Completions API `reasoning_effort` `medium` maps to `thinking_level` `high`.  
  OpenAI Chat Completions API 中的 `reasoning_effort` 为 `medium` 时，对应 `thinking_level` 的 `high`。

- Thinking can't be turned off for Gemini 3.1 Pro.  
  在 Gemini 3.1 Pro 中无法关闭思考功能。

````
prompt = """
你的任务是使用现代 C++ 实现经典的线程安全双重检查锁定（DCL，Double-Checked Locking）单例模式。该任务并非易事，需要具备专业的并发编程知识以防止出现内存重排（memory reordering）问题。
请编写一个名为 dcl_singleton.cpp 的完整且可运行的 C++ 程序，在其中定义一个具有私有构造函数和静态 getInstance() 方法的 Singleton 类。
你的解决方案必须严格遵守以下约束条件：
1.单例的实例指针（static Singleton*）必须封装在 std::atomic 中，以便正确管理跨线程的内存可见性。
2.getInstance() 方法在进行外层检查（读取实例指针）时，必须使用 std::memory_order_acquire。
3.在创建实例并将其写回（赋值给）原子指针时，必须使用 std::memory_order_release。
4.必须使用标准的 std::mutex，且仅用于保护临界区（即实际执行实例化的代码部分）。
m5.ain 函数必须演示安全的并发访问：启动至少三个线程，每个线程都调用 Singleton::getInstance()，并打印返回实例的内存地址，以证明所有线程获取到的都是同一个对象。
"""

response = client.models.generate_content(
    model=MODEL_ID,
    contents=prompt,
    config=types.GenerateContentConfig(
        thinking_config=types.ThinkingConfig(
            thinking_level=types.ThinkingLevel.HIGH  # Dynamic thinking for high reasoning tasks
        )
    ),
)

display(Markdown(response.text))
````

### 2️⃣ Media Resolution  
### 2️⃣ 媒体分辨率（Media Resolution）

Gemini 3 introduces granular control over multimodal vision processing via the `media_resolution` parameter. Higher resolutions improve the model's ability to read fine text or identify small details, but increase token usage and latency. The `media_resolution` parameter determines the maximum number of tokens allocated per input image or video frame.  
Gemini 3 通过 `media_resolution` 参数引入了对多模态视觉处理的精细化控制。更高的分辨率可以提升模型读取细小文字或识别细节的能力，但同时也会增加 token 使用量和延迟。`media_resolution` 参数用于决定每个输入图像或视频帧可分配的最大 token 数量。

**⚠️ Note**  
**⚠️ 注意**

Because media resolution directly impacts token count, you may need to lower the resolution (e.g., to `low`) to fit very long inputs, such as long videos or extensive documents.  
由于媒体分辨率会直接影响 token 数量，对于非常长的输入（例如长视频或大型文档），您可能需要降低分辨率（例如设置为 `low`）以满足限制。

You can set the resolution to `low`, `medium`, or `high` per individual media part, as example:  
您可以为每个独立的媒体部分单独设置分辨率为 `low`、`medium` 或 `high`，例如：

````
response = client.models.generate_content(
    model=MODEL_ID,
    contents=[
        types.Part(
            file_data=types.FileData(
                file_uri="gs://cloud-samples-data/generative-ai/image/a-man-and-a-dog.png",
                mime_type="image/jpeg",
            ),
            media_resolution=types.PartMediaResolution(
                level=types.PartMediaResolutionLevel.MEDIA_RESOLUTION_HIGH
            ),
        ),
        types.Part(
            file_data=types.FileData(
                file_uri="gs://cloud-samples-data/generative-ai/video/behind_the_scenes_pixel.mp4",
                mime_type="video/mp4",
            ),
            media_resolution=types.PartMediaResolution(
                level=types.PartMediaResolutionLevel.MEDIA_RESOLUTION_LOW
            ),
        ),
        "图像什么时候出现在视频中？背景是什么？",
    ],
)

display(Markdown(response.text))
````

Or set it globally via `GenerateContentConfig`. If unspecified, the model uses optimal defaults based on the media type.

````
response = client.models.generate_content(
    model=MODEL_ID,
    contents=[
        types.Part(
            file_data=types.FileData(
                file_uri="gs://cloud-samples-data/generative-ai/image/a-man-and-a-dog.png",
                mime_type="image/jpeg",
            ),
        ),
        "图片中是什么?",
    ],
    config=types.GenerateContentConfig(
        media_resolution=types.MediaResolution.MEDIA_RESOLUTION_LOW
    ),
)

display(Markdown(response.text))
````

### 3️⃣ Thought Signature  
### 3️⃣ 思考签名（Thought Signature）

Thought signatures are encrypted tokens that preserve the model's reasoning state during multi-turn conversations, specifically when using Function Calling.  
思考签名（Thought Signature）是一种加密 token，用于在多轮对话中（尤其是在使用函数调用时）保留模型的推理状态。

When a thinking model decides to call an external tool, it pauses its internal reasoning process. The thought signature acts as a "save state", allowing the model to resume its chain of thought seamlessly once you provide the function's result.  
当思考型模型决定调用外部工具时，会暂停其内部推理过程。此时，思考签名就像一个“存档点”，在您提供函数执行结果后，模型可以无缝地继续之前的思考链路。

Gemini 3 enforces stricter validation and updated handling on thought signatures which were originally introduced in Gemini 2.5. To ensure the model maintains context across multiple turns of a conversation, you must return the thought signatures in your subsequent requests.  
Gemini 3 对最初在 Gemini 2.5 中引入的思考签名进行了更严格的校验和更新的处理方式。为了确保模型在多轮对话中保持上下文一致，您必须在后续请求中返回这些思考签名。


#### Automatic Handling of Thought Signatures (Recommended)  
#### 思考签名的自动处理（推荐）

If you are using the Google Gen AI SDKs (Python, Node.js, Go, Java) or OpenAI Chat Completions API, and utilizing the standard chat history features or appending the full model response, thought signatures are handled automatically. You do not need to make any changes to your code.  
如果您使用的是 Google Gen AI SDK（Python、Node.js、Go、Java）或 OpenAI Chat Completions API，并且使用标准的对话历史记录功能或追加完整的模型响应，那么思考签名会被自动处理，您无需对代码进行任何修改。


#### **Example 1**: Automatic Function Calling

*   列表项
*   列表项



When using the Gen AI SDK in automatic function calling, thought signatures are handled automatically.  
在使用 Gen AI SDK 的自动函数调用（automatic function calling）时，思考签名会被自动处理。

````
def get_weather(city: str) -> str:
    """Gets the weather in a city."""
    if "london" in city.lower():
        return "Rainy"
    if "new york" in city.lower():
        return "Sunny"
    return "Cloudy"


response = client.models.generate_content(
    model=MODEL_ID,
    contents="What's the weather in London and New York?",
    config=types.GenerateContentConfig(
        tools=[get_weather],
    ),
)

# The SDK handles the function calls and thought signatures, and returns the final text
print("Final response:", response.text)

# Print function calling history
hist_turn = response.automatic_function_calling_history[1]
print("\nFunction Call 1:", hist_turn.parts[1].function_call.name)
````

#### **Example 2**: Manual Function Calling

When using the Gen AI SDK in manual function calling, thought signatures are also handled automatically if you append the full model response in sequential model requests.  
在使用 Gen AI SDK 进行手动函数调用（manual function calling）时，如果在连续的模型请求中追加完整的模型响应，思考签名同样会被自动处理。

````
# 1. Define your tool
get_weather_declaration = types.FunctionDeclaration(
    name="get_weather",
    description="Gets the current weather temperature for a given location.",
    parameters={
        "type": "object",
        "properties": {"location": {"type": "string"}},
        "required": ["location"],
    },
)
get_weather_tool = types.Tool(function_declarations=[get_weather_declaration])

# 2. Send a message that triggers the tool
prompt = "What's the weather like in London?"
response = client.models.generate_content(
    model=MODEL_ID,
    contents=prompt,
    config=types.GenerateContentConfig(
        tools=[get_weather_tool],
        thinking_config=types.ThinkingConfig(include_thoughts=True),
    ),
)

# 3. Handle the function call
function_call = response.function_calls[0]
location = function_call.args["location"]
print(f"Model wants to call: {function_call.name}")

# Execute your tool (e.g., call an API)
# (This is a mock response for the example)
print(f"Calling external tool for: {location}")
function_response_data = {
    "location": location,
    "temperature": "30C",
}

# 4. Send the tool's result back
# Append this turn's messages to history for a final response.
# The `content` object automatically attaches the required thought_signature behind the scenes.
history = [
    types.Content(role="user", parts=[types.Part(text=prompt)]),
    response.candidates[0].content,  # Signature preserved here
    types.Content(
        role="tool",
        parts=[
            types.Part.from_function_response(
                name=function_call.name,
                response=function_response_data,
            )
        ],
    ),
]

response_2 = client.models.generate_content(
    model=MODEL_ID,
    contents=history,
    config=types.GenerateContentConfig(
        tools=[get_weather_tool],
        thinking_config=types.ThinkingConfig(include_thoughts=True),
    ),
)

# 5. Get the final, natural-language answer
print(f"\nFinal model response: {response_2.text}")
````

#### Manual Handling of Thought Signatures  
#### 思考签名的手动处理

If you are interacting with the API directly or managing raw JSON payloads, you must correctly handle the `thought_signature` included in the model's turn.  
如果您是直接与 API 交互或手动管理原始 JSON 请求数据，则必须正确处理模型响应中包含的 `thought_signature`。

You must return this signature in the exact part where it was received when sending the conversation history back.
在发送对话历史时，必须在接收到该签名的**相同位置**原样返回该 `thought_signature`。

⚠️ If proper signatures are not returned, the model will return a 400 Error `"Function Call in the content block is missing a thought_signature"`.  
⚠️ 如果未正确返回该签名，模型将返回 400 错误：`"Function Call in the content block is missing a thought_signature"`。

See [documentation](https://docs.cloud.google.com/vertex-ai/generative-ai/docs/multimodal/function-calling#thinking) for more details.  
更多信息请参阅[官方文档](https://docs.cloud.google.com/vertex-ai/generative-ai/docs/multimodal/function-calling#thinking)。

### 4️⃣ Streaming Function Calling

You can use streaming partial function call arguments to improve streaming experience on tool use. This feature can be enabled by explicitly setting `stream_function_call_arguments` to true.  
您可以使用函数调用参数的流式分块（streaming partial function call arguments）来提升工具调用时的流式体验。可以通过显式将 `stream_function_call_arguments` 设置为 `true` 来启用该功能。

````
get_weather_declaration = types.FunctionDeclaration(
    name="get_weather",
    description="Gets the current weather temperature for a given location.",
    parameters={
        "type": "object",
        "properties": {"location": {"type": "string"}},
        "required": ["location"],
    },
)
get_weather_tool = types.Tool(function_declarations=[get_weather_declaration])


for chunk in client.models.generate_content_stream(
    model=MODEL_ID,
    contents="What's the weather in London and New York?",
    config=types.GenerateContentConfig(
        tools=[get_weather_tool],
        tool_config=types.ToolConfig(
            function_calling_config=types.FunctionCallingConfig(
                mode=types.FunctionCallingConfigMode.AUTO,
                stream_function_call_arguments=True,
            )
        ),
    ),
):
    function_call = chunk.function_calls[0]
    if function_call and function_call.name:
        print(f"{function_call.name}")
        print(f"will_continue={function_call.will_continue}")
````

### 5️⃣ Multimodal Function Responses  
### 5️⃣ 多模态函数响应（Multimodal Function Responses）

Multimodal function calling allows users to have function responses containing multimodal objects allowing for improved utilization of function calling capabilities of the model.  
多模态函数调用允许用户在函数响应中包含多模态对象，从而更充分地利用模型的函数调用能力。

Currently function calling only supports text based function responses.  
目前，函数调用仅支持基于文本的函数响应。

````
# 1. Define the function tool
get_image_declaration = types.FunctionDeclaration(
    name="get_image",
    description="Retrieves the image file reference for a specific order item.",
    parameters={
        "type": "object",
        "properties": {
            "item_name": {
                "type": "string",
                "description": "The name or description of the item ordered (e.g., 'green shirt').",
            }
        },
        "required": ["item_name"],
    },
)
tool_config = types.Tool(function_declarations=[get_image_declaration])

# 2. Send a message that triggers the tool
prompt = "Show me the green shirt I ordered last month."
response_1 = client.models.generate_content(
    model=MODEL_ID,
    contents=[prompt],
    config=types.GenerateContentConfig(
        tools=[tool_config],
    ),
)

# 3. Handle the function call
function_call = response_1.function_calls[0]
requested_item = function_call.args["item_name"]
print(f"Model wants to call: {function_call.name}")

# Execute your tool (e.g., call an API)
# (This is a mock response for the example)
print(f"Calling external tool for: {requested_item}")

function_response_data = {
    "image_ref": {"$ref": "dress.jpg"},
}

function_response_multimodal_data = types.FunctionResponsePart(
    file_data=types.FunctionResponseFileData(
        mime_type="image/png",
        display_name="dress.jpg",
        file_uri="gs://cloud-samples-data/generative-ai/image/dress.jpg",
    )
)

# 4. Send the tool's result back
# Append this turn's messages to history for a final response.
history = [
    types.Content(role="user", parts=[types.Part(text=prompt)]),
    response_1.candidates[0].content,
    types.Content(
        role="tool",
        parts=[
            types.Part.from_function_response(
                name=function_call.name,
                response=function_response_data,
                parts=[function_response_multimodal_data],
            )
        ],
    ),
]

response_2 = client.models.generate_content(
    model=MODEL_ID,
    contents=history,
    config=types.GenerateContentConfig(
        tools=[tool_config],
        thinking_config=types.ThinkingConfig(include_thoughts=True),
    ),
)

print(f"\nFinal model response: {response_2.text}")
````

## ⭐️ Supported Features

### ✅ Safety Filters

````
system_instruction = "Be as mean and hateful as possible."

prompt = """
Write a list of 5 disrespectful things that I might say to the universe after stubbing my toe in the dark.
"""

safety_settings = [
    types.SafetySetting(
        category=types.HarmCategory.HARM_CATEGORY_DANGEROUS_CONTENT,
        threshold=types.HarmBlockThreshold.BLOCK_LOW_AND_ABOVE,
    ),
    types.SafetySetting(
        category=types.HarmCategory.HARM_CATEGORY_HARASSMENT,
        threshold=types.HarmBlockThreshold.BLOCK_LOW_AND_ABOVE,
    ),
    types.SafetySetting(
        category=types.HarmCategory.HARM_CATEGORY_HATE_SPEECH,
        threshold=types.HarmBlockThreshold.BLOCK_LOW_AND_ABOVE,
    ),
    types.SafetySetting(
        category=types.HarmCategory.HARM_CATEGORY_SEXUALLY_EXPLICIT,
        threshold=types.HarmBlockThreshold.BLOCK_LOW_AND_ABOVE,
    ),
]

response = client.models.generate_content(
    model=MODEL_ID,
    contents=prompt,
    config=types.GenerateContentConfig(
        system_instruction=system_instruction,
        safety_settings=safety_settings,
    ),
)

# Response will be `None` if it is blocked.
print(response.text)
# Finish Reason will be `SAFETY` if it is blocked.
print(response.candidates[0].finish_reason)
# Safety Ratings show the levels for each filter.
for safety_rating in response.candidates[0].safety_ratings:
    print(safety_rating)
````

### ✅ Search as a Tool

Gemini 3.1 includes native support for formatting citations directly in the text (e.g., `Spain won Euro 2024... [1](https://...)`).

````
response = client.models.generate_content(
    model=MODEL_ID,
    contents="Where will the next FIFA World Cup be held?",
    config=types.GenerateContentConfig(
        tools=[types.Tool(google_search=types.GoogleSearch())],
    ),
)
````

Use this helper function to print the full citations.

````
def print_grounding_data(response: types.GenerateContentResponse) -> None:
    """Print Gemini response with grounding citations in Markdown format."""
    candidate = response.candidates[0] if response.candidates else None
    metadata = getattr(candidate, "grounding_metadata", None)

    if not metadata:
        print("Response does not contain grounding metadata.")
        display(Markdown(response.text))
        return

    ENCODING = "utf-8"
    text_bytes = response.text.encode(ENCODING)
    parts = []
    last = 0

    # Insert citation markers into text
    for support in metadata.grounding_supports or []:
        end = support.segment.end_index
        parts.append(text_bytes[last:end].decode(ENCODING))
        parts.append(
            " " + "".join(f"[{i + 1}]" for i in support.grounding_chunk_indices)
        )
        last = end

    parts.append(text_bytes[last:].decode(ENCODING))
    parts.append("\n\n----\n## Grounding Sources\n")

    # Grounding chunks
    if chunks := metadata.grounding_chunks:
        parts.append("### Grounding Chunks\n")
        for i, chunk in enumerate(chunks, 1):
            if not (ctx := chunk.web or chunk.retrieved_context or chunk.maps):
                continue
            uri = (
                (ctx.uri or "")
                .replace("gs://", "https://storage.googleapis.com/", 1)
                .replace(" ", "%20")
            )
            parts.append(f"{i}. [{ctx.title or 'Source'}]({uri})\n")
            if getattr(ctx, "place_id", None):
                parts.append(f"    - Place ID: `{ctx.place_id}`\n\n")
            if getattr(ctx, "text", None):
                parts.append(f"{ctx.text}\n\n")

    # Queries
    if metadata.web_search_queries:
        parts.append(f"\n**Web Search Queries:** {metadata.web_search_queries}\n")
        if metadata.search_entry_point:
            parts.append(
                f"\n**Search Entry Point:**\n{metadata.search_entry_point.rendered_content}\n"
            )
    elif metadata.retrieval_queries:
        parts.append(f"\n**Retrieval Queries:** {metadata.retrieval_queries}\n")

    display(Markdown("".join(parts)))


print_grounding_data(response)
````
